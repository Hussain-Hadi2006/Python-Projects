```python
import string
lowercase_alphabet=string.ascii_lowercase+string.ascii_lowercase
uppercase_alphabet=string.ascii_uppercase+string.ascii_uppercase
def caesar_encrypt():
  word=input("please type a word: ")
  number=int(input("Please type the shift number: "))
  encrypted_word=""
  for i in word:
    if i in lowercase_alphabet:
      orig_pos=lowercase_alphabet.index(i)
      new_posi=orig_pos+number
      encrypted_word +=lowercase_alphabet[new_posi]
    elif i in uppercase_alphabet:
      orig_pos=uppercase_alphabet.index(i)
      new_posi=orig_pos+number
      encrypted_word +=uppercase_alphabet[new_posi]
    else:
      encrypted_word +=i
  print(encrypted_word)
def caesar_decrypt():
  word=input("please type a encrypt_word: ")
  number=int(input("Please type the shift number: "))
  decrypted_word=""
  for i in word:
    if i in lowercase_alphabet:
      orig_pos=lowercase_alphabet.index(i)
      new_posi=orig_pos-number
      decrypted_word +=lowercase_alphabet[new_posi]
    elif i in uppercase_alphabet:
      orig_pos=uppercase_alphabet.index(i)
      new_posi=orig_pos-number
      decrypted_word +=uppercase_alphabet[new_posi]
    else:
      decrypted_word +=i
  print(decrypted_word)
def caesar_cipher():
  choice=input("Do you want to (encrypt) or (decrypt): ").lower()
  if choice=="encrypt":
    caesar_encrypt()
  elif choice=="decrypt":
    caesar_decrypt()
  else:
    print("Please choice (encrypt) or (decrypt)")
caesar_cipher
```
