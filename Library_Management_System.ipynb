```python
def add_book():
  while True:
    isbn=int(input("Enter ISBN: "))
    title=input("Enter title: ")
    author=input("Enter auther: ")
    information[isbn]={'title':title,"author":author,"Available":True}
    print(f"Book {title} by {author} added to the catalog with ISBN {isbn}")
    choose=input("Do you want add another book? (y/n)").lower()
    if choose=="n":
      break
def list_books():
  while True:
    print("\nLibrary catalog")
    for isbn in information:
      book_info=information[isbn]
      print(f"ISBN:{isbn}, Title:{book_info["title"]}, Auther:{book_info["author"]}, Available:{book_info["Available"]}")
    choice=input("DO you want to go back to the main menu? (y/n)")
    if choice.lower()=="y":
      break
def check_out():
  while True:
    isbn=(input("Enter ISBN to check out: "))
    if isbn in information:
      if information[isbn]["Available"]:
        information[isbn]["Available"]=False
        print(f"Book '{information[isbn]['title']}' checked out successfully.")
      else:
        print("Sorry, the book is currently checked out.")
    else:
      print("Book not found in the catalog.")
    choose_out=input("Do you want to check out another book? (y/n)")
    if choose_out.lower()!="y":
      break

def check_in():
  while True:
    isbn=(input("Enter ISBN to check out: "))
    if isbn in information:
      if not information[isbn]["Available"]:
        information[isbn]["Available"]=True
      else:
        print("The book is already available in the library.")
    else:
      print("Book not found in the catalog.")
    choose_in=input("Do you want to check in another book? (y/n)")
    if choose_in.lower()!="y":
      break
information={}
while True:
  print("""
           Menu:
           1.Add Book
           2.Check Out Book
           3.Check In Book
           4.List Books
           5.Exit
            """)
  choice_number=(input("Enter your choice (1-5):"))
  if choice_number=="1":
    add_book()
  elif choice_number=="2":
    check_out()
  elif choice_number=="3":
    check_in()
  elif choice_number=="4":
    list_books()
  elif choice_number=="5":
    break
```
