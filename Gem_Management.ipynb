```python
class Gym():
    def __init__(self, first_name,last_name,membership_id,status="inactive"):
        self.first_name=first_name
        self.last_name=last_name
        self.membership_id=membership_id
        self.status=status
    def display(self):
        print(f'first name: {self.first_name}')
        print(f'last_name: {self.last_name}')
        print(f'membership_id: {self.membership_id}')
        print(f'status: {self.status}')
        print("_" * 20)
def create_membership():
    first_name=input('Enter first name:  ')
    last_name=input('Enter last name:  ')
    membership_id=input('Enter membership ID:  ')
    status=input('Enter membership status (active/inactive), Or click enter:  ')
    while True:
        if len(status)==0:
            status="inactive"
            break
        elif status not in ["active","inactive"]:
            print("sorry,this wrong")
            status=input('Enter membership status (active/inactive), Or click enter:  ')
        else:
            break
    return Gym(first_name,last_name,membership_id,status)
def search(members):
    print("\nSearch by: \n")
    print("1. membership ID")
    print("2. First Name")
    print("3. Membership Status\n")
    search_choice=input("Enter your choice:  ")
    found_members=[]
    if search_choice=="1":
        search_id=input("Enter the membership ID to search:  ")
        for x in members:
            if x.membership_id==search_id:
                found_members.append(x)
                break

    elif search_choice=="2":
        search_first_name=input("Enter the first name to search:  ")
        for x in members:
            if x.first_name.lower()==search_first_name.lower():
                found_members.append(x)
    elif search_choice=="3":
        search_status=input("Enter the membership status to search:  ")
        for x in members:
            if x.status.lower()==search_status.lower():
                found_members.append(x)
    else:
        print("Wrong choice, please choice from (1-3).")
    if found_members:
        print('Members found:  ')
        for x in found_members:
            x.display()
    else:
        print("Member not found!")

members=[]
while True:
    print('Welcom to Gem membership managment \n\n\n')
    print('''Choose an Action:

1-Add new member
2-Display all members
3-serach for a member
4-Exit''')
    choose=input('\n\nEntre your choice:  ')
    if choose=="1":
        members.append(create_membership())
        print("Member added successfully!!")
    elif choose=="2":
        if members:
            print("Display all members....")
            for i in members:
                i.display()
        else:
            print("No members to display")
    elif choose=="3":
        if members:
            search(members)
        else:
            print("No members to search!")
    elif choose=="4":
        print("Exiting....")
        break
    else:
        print("Wrong choice, please choice from (1-4).")
```
