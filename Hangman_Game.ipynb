```python
import random
past_guesses=[]
HANGMANPICS = ['''
  +---+
  |   |
      |
      |
      |
      |
=========''',r'''
  +---+
  |   |
  O   |
      |
      |
      |
=========''',r'''
  +---+
  |   |
  O   |
  |   |
      |
      |
=========''',r'''
  +---+
  |   |
  O   |
 /|   |
      |
      |
=========''',r'''
  +---+
  |   |
  O   |
 /|\  |
      |
      |
=========''',r'''
  +---+
  |   |
  O   |
 /|\  |
 /    |
      |
=========''',r'''
  +---+
  |   |
  O   |
 /|\  |
 / \  |
      |
========='''][::-1]
words = ["good", "bad", "ugly"]
random_word = random.choice(words)
display = ["_"]*len(random_word)
lives=7
print(" ".join(display))
print("""
  +---+
  |   |
      |
      |
      |
      |
=========""")
while "_" in display:
   guessed = input("\nأختر حرف يا بطاطس: ").lower()
   if guessed in past_guesses:
    print("لقد اخترت هذا الحرف من قبل ,حاول مجددا")
    continue
   past_guesses.append(guessed)
   for position in range(len(random_word)):
    if random_word[position]==guessed:
      display[position]= guessed
   if "_" not in display:
    print("\nلقد فزت يا بطاطس")
    print(' '.join(display))
    break
   if  guessed not in random_word:
    lives=lives-1
    print(HANGMANPICS[lives])
    print(f"\nتبقى لديك({lives})محاولات يا بطاطس ")
    print("أبذل جهدك أنت تستطيع")
   print(' '.join(display))
   if lives==0:
    print("\nانت مسكين يا بطاطس لقد خسرت,حظا اوفرالمرة القادمة")
    break
```
